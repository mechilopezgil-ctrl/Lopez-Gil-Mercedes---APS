## Ecuación primera

La primera ecuación con la que trabajaremos es y(n)=x(n−3)+x(n−2)+x(n−1)+x(n).
 

<div style="text-align: center;">
  <img src="TS6_ejA.jpeg" width="750">
</div>

Ya habiendo calculado manualmente y graficado la respuesta en frecuencia tanto del módulo como de la fase, verificamos los resultados obtenidos en el sistema utilizando Numpy.

```python
import numpy as np
import matplotlib.pyplot as plt

# coeficientes b_i 
b = [1, 1, 1, 1]

# Vector de frecuencias
omega = np.linspace(0, np.pi, 512)

T = np.zeros(len(omega), dtype=complex)
for i, bi in enumerate(b):
    T += bi * np.exp(-1j * omega * i)

# Módulo y fase
modulo = np.abs(T)
fase = np.angle(T)


plt.subplot(2,1,1)
plt.plot(omega, modulo)
plt.title("Módulo")

plt.subplot(2,1,2)
plt.plot(omega, fase)
plt.title("Fase")
plt.tight_layout()
plt.show()
```

## Ejercicio cuarto

La cuarta ecuación con la que trabajaremos es y(n)=x(n)−x(n−2).


<div style="text-align: center;">
  <img src="TS6_ejD.jpeg" width="750">
</div>

```python

# Tus coeficientes b_i (ejemplo)
b = [1, 0, -1]

# Vector de frecuencias
omega = np.linspace(0, np.pi, 512)

# DTFT: suma de b_i * e^(-j*omega*i)
T = np.zeros(len(omega), dtype=complex)
for i, bi in enumerate(b):
    T += bi * np.exp(-1j * omega * i)

# Módulo y fase
modulo = np.abs(T)
fase = np.angle(T)


plt.subplot(2,1,1)
plt.plot(omega, modulo)
plt.title("Módulo")

plt.subplot(2,1,2)
plt.plot(omega, fase)
plt.title("Fase")
plt.tight_layout()
plt.show()
```
